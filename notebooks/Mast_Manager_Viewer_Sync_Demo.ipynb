# Mast Manager Viewer Sync Demo

```python
from mast_aladin import MastAladin, MastManager
from sidecar import Sidecar
import jdaviz as jd
```

### Setup
Create an instance of the MastManager class 

```python
mm = MastManager()
```

Create an instance of MastAladin

```python
mast_aladin = MastAladin(target="Cartwheel Galaxy")

sc = Sidecar(title="mast-aladin", anchor='split-right')
with sc:
    display(mast_aladin)

mm.register_app(mast_aladin, "mast-aladin")
```

Create an instance of Jdaviz

```python
jd.load("mast:JWST/product/jw02727-o002_t062_nircam_clear-f090w_i2d.fits", format="Image", cache=True)
```

```python
sc = Sidecar(title="jdaviz Main", anchor='split-bottom', ref=sc)

with sc:
    jd.show()

mm.register_app(jd.gca(), "viz")
```

### Viewer Sync Plugin

```python
mm.display_plugin('viewer_sync')
```
