# Mast Manager Viewer Sync Demo

```python
from mast_aladin import MastAladin, MastManager
from sidecar import Sidecar
```

### Setup
Create an instance of the MastManager class 

```python
mm = MastManager()
```

Create two instances of MastAladin (a, b)

```python
mast_aladin_a = MastAladin()

sc = Sidecar(title="mast-aladin-a", anchor='split-right')
with sc:
    display(mast_aladin_a)

mm.register_app(mast_aladin_a, "mast-aladin-a")
```

```python
mast_aladin_b = MastAladin()

sc = Sidecar(title="mast-aladin-b", anchor='split-bottom', ref=sc)
with sc:
    display(mast_aladin_b)

mm.register_app(mast_aladin_b, "mast-aladin-b")
```

### Viewer Sync Plugin

```python
mm.display_plugin('viewer_sync')
```
